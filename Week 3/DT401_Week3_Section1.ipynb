### Understanding the Six Dimensions

Data quality measures whether data serves its intended purpose. Each dimension addresses a specific aspect:

**1. Validity**: Does data conform to expected format, type, and range?  
*Example*: Temperature readings of 150°C in UK weather data are invalid.

**2. Uniqueness**: Is data free from unexpected duplication?  
*Example*: The same weather observation recorded twice creates duplicate records.

**3. Completeness**: Is required information present?  
*Example*: Weather data missing temperature readings can't support climate analysis.

**4. Timeliness**: Is data available when needed?  
*Example*: Receiving yesterday's weather forecast is useless for today's planning.

**5. Accuracy**: Does data reflect reality?  
*Example*: A weather station reporting 20°C when actual temperature is 5°C produces inaccurate data.

**6. Consistency**: Do values align across records and systems?  
*Example*: The same location showing different coordinates in different records indicates inconsistency.

<div class="alert alert-block alert-warning">
<b>Reminder:</b> <br>
These six dimensions apply universally across all data types and sectors. Whether you're working with patient records, financial transactions, or environmental data, the same quality framework applies.
</div>

---

### Understanding Completeness Checks

Completeness assesses whether required information is present. Missing data can occur due to:
- Sensor failures or collection errors
- Data transmission issues
- System integration gaps
- User input omissions

Python provides several methods to check completeness:
- `.isnull()` - returns True for missing values
- `.sum()` - counts True values (missing entries)
- `.isna()` - synonym for `.isnull()`

<div class="alert alert-block alert-warning">
<b>Reminder:</b> <br>
Different business contexts have different completeness requirements. Critical safety data requires 100% completeness; optional survey responses may accept lower completeness rates.
</div>

### Worked Example: Calculating Completeness Scores

````python
# Calculate completeness percentage
total_cells = df.shape[0] * df.shape[1]
missing_cells = df.isnull().sum().sum()
completeness = ((total_cells - missing_cells) / total_cells) * 100

print(f"Total cells: {total_cells:,}")
print(f"Missing cells: {missing_cells:,}")
print(f"Overall completeness: {completeness:.2f}%")
````

<div class="alert alert-block alert-info">
<b>What just happened?</b><br>

- We calculated total cells by multiplying rows by columns
- We counted missing cells using `.isnull().sum().sum()`
- We calculated completeness as percentage of non-missing cells
- A score of 100% means no missing data; lower scores indicate gaps

**Key lesson:** Overall completeness gives a high-level quality indicator, but column-level analysis reveals where problems exist.

</div>

Now check which columns have missing values:

````python
# Missing values per column
missing_by_column = df.isnull().sum()
missing_with_percent = pd.DataFrame({
    'Missing_Count': missing_by_column,
    'Missing_Percent': (missing_by_column / len(df)) * 100
})

# Show only columns with missing values
print(missing_with_percent[missing_with_percent['Missing_Count'] > 0])
````

**What this diagram shows:**

**Why this matters in your role:**

---

### Understanding Validity Checks

Validity assesses whether data conforms to expected formats and ranges. For weather data:
- Temperature: -50°C to 50°C (reasonable for Earth)
- Humidity: 0% to 100%
- Wind speed: 0 to 200 mp/h (beyond hurricane force is suspicious)
- Pressure: 950 to 1050 hPa (normal atmospheric range)

Python allows conditional checks:
```python
# Check if values are within range
invalid_rows = df[(df['column'] < min_value) | (df['column'] > max_value)]
```

<div class="alert alert-block alert-warning">
<b>Reminder:</b> <br>
Validity ranges depend on context. A heart rate of 200 bpm is invalid for resting measurements but valid during intense exercise. Always define ranges based on business context.
</div>

### Worked Example: Checking Temperature Validity

````python
# Define valid temperature range for UK
temp_min = -20
temp_max = 40

# Check for invalid temperatures
invalid_temp = df[(df['temperature_2m (°C)'] < temp_min) | (df['temperature_2m (°C)'] > temp_max)]

print(f"Total observations: {len(df)}")
print(f"Invalid temperatures: {len(invalid_temp)}")
print(f"Validity rate: {((len(df) - len(invalid_temp)) / len(df)) * 100:.2f}%")
````

<div class="alert alert-block alert-info">
<b>What just happened?</b><br>

- We defined acceptable temperature range for UK climate (-20°C to 40°C)
- We used boolean indexing to find values outside this range
- We calculated validity rate as percentage of valid observations
- The `|` operator means "OR" - flagging values below minimum OR above maximum

**Key lesson:** Validity checks require domain knowledge. These ranges work for UK weather but would differ for Death Valley or Antarctica.

</div>

Check if any invalid temperatures exist:

````python
if len(invalid_temp) > 0:
    print("\nInvalid temperature examples:")
    print(invalid_temp[['time', 'temperature_2m (°C)']].head())
else:
    print("\nGood news: All temperatures are within valid range!")
````

---

### Understanding Uniqueness Checks

Uniqueness assesses whether records are free from unexpected duplication. Duplicates can occur through:
- Data loading errors (loading same file twice)
- System integration issues (multiple systems creating same record)
- User actions (submitting same form multiple times)

For time series data like weather observations, each timestamp should appear only once. Duplicates indicate collection or storage issues.

Python provides methods to check uniqueness:
- `.duplicated()` - returns True for duplicate rows
- `.drop_duplicates()` - removes duplicate rows
- `.nunique()` - counts unique values

<div class="alert alert-block alert-warning">
<b>Reminder:</b> <br>
Some apparent duplicates are valid. Multiple customers can share the same name or address. Check whether duplication violates business rules before removing records.
</div>

### Worked Example: Checking for Duplicate Timestamps

````python
# Check for duplicate timestamps
duplicate_times = df[df.duplicated('time', keep=False)]

print(f"Total observations: {len(df)}")
print(f"Duplicate timestamps: {len(duplicate_times)}")
print(f"Uniqueness rate: {((len(df) - len(duplicate_times)) / len(df)) * 100:.2f}%")
````

<div class="alert alert-block alert-info">
<b>What just happened?</b><br>

- We used `.duplicated()` to find rows with duplicate 'time' values
- The `keep=False` parameter marks ALL duplicates (not just first or last occurrence)
- We calculated uniqueness rate as percentage of non-duplicate records
- For time series data, 100% uniqueness is expected

**Key lesson:** Time series data should have unique timestamps. Duplicates indicate data collection issues requiring investigation.

</div>

Show duplicate examples if any exist:

````python
if len(duplicate_times) > 0:
    print("\nDuplicate timestamp examples:")
    print(duplicate_times[['time', 'temperature_2m (°C)']].sort_values('time').head(10))
else:
    print("\nGood news: All timestamps are unique!")
````


**What this diagram shows:**  
A comprehensive quality report combines results from multiple dimension checks into actionable recommendations.

**Why this matters in your role:**  
Stakeholders need clear quality reports to make decisions. A report stating "95% complete, 99% valid, 100% unique" helps managers decide if data is fit for purpose or needs improvement before critical analysis.

---

### Understanding Quality Reports

A quality report summarizes findings across all dimensions, providing:
- Quantitative metrics (percentages, counts)
- Qualitative assessment (fit for purpose decision)
- Actionable recommendations (what to fix first)
- Business impact (how quality affects decisions)

Reports should be concise, focused on business implications rather than technical details.

<div class="alert alert-block alert-warning">
<b>Reminder:</b> <br>
Quality requirements vary by use case. Real-time fraud detection needs 100% timeliness; historical analysis can tolerate some delay. Tailor quality standards to business needs.
</div>

### Worked Example: Building a Quality Summary

````python
# Calculate quality metrics
total_records = len(df)
completeness_pct = ((df.shape[0] * df.shape[1] - df.isnull().sum().sum()) / (df.shape[0] * df.shape[1])) * 100
uniqueness_pct = ((total_records - len(df[df.duplicated('time')])) / total_records) * 100

# Display summary
print("=== DATA QUALITY SUMMARY ===")
print(f"\nDataset: Weather Observations")
print(f"Records: {total_records:,}")
print(f"Completeness: {completeness_pct:.2f}%")
print(f"Uniqueness: {uniqueness_pct:.2f}%")
````

<div class="alert alert-block alert-info">
<b>What just happened?</b><br>

- We calculated completeness by counting non-missing cells
- We calculated uniqueness by checking timestamp duplication
- We formatted output as clear, readable summary
- This structure is suitable for presenting to stakeholders

**Key lesson:** Quality reports should be stakeholder-friendly. Focus on percentages and clear categories rather than technical details.

</div>
