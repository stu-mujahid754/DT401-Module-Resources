## Part 1: Understanding Data Gathering Approaches

### Visual Overview

```mermaid
flowchart TD
    A[Data Gathering Decision] --> B{What do you need to know?}
    B --> C[WHAT is happening?]
    B --> D[WHY is it happening?]
    C --> E[Quantitative Methods]
    D --> F[Qualitative Methods]
    E --> G[Numbers, counts, measurements]
    F --> H[Experiences, opinions, context]
    G --> I[Surveys, experiments, secondary data]
    H --> J[Interviews, focus groups, observations]
```

**What this diagram shows:**
Data gathering begins with understanding your question. Quantitative methods answer "what" questions with numbers; qualitative methods answer "why" questions with context.

**Why this matters in your role:**
Choosing the wrong method wastes time and produces unhelpful data. A healthcare trust asking "why" patients miss appointments won't find answers in attendance numbers alone—they need qualitative interviews to understand barriers.

---

### Quantitative vs. Qualitative: A Comparison

Let's use weather data to demonstrate different gathering approaches.

````python
import pandas as pd
import numpy as np

# Load weather dataset (Open-Meteo API, London UK, 2023-2025)
weather_df = pd.read_csv('../open-meteo-51.49N0.16W23m.csv', skiprows=3, low_memory=False)
weather_df.columns = [c.split(' ')[0].strip() for c in weather_df.columns]
weather_df = weather_df.rename(columns={
    'time': 'date',
    'temperature_2m': 'temperature',
    'wind_speed_10m': 'wind_speed',
    'relative_humidity_2m': 'relative_humidity'
})
weather_df = weather_df[weather_df['date'].str.match(r'^\d{4}-\d{2}-\d{2}T', na=False)].reset_index(drop=True)
for col in weather_df.columns[1:]:
    weather_df[col] = pd.to_numeric(weather_df[col], errors='coerce')

# Display first few rows
print("Weather Dataset - First 5 Observations:")
print(weather_df.head())
````

<div class="alert alert-block alert-warning">
<b>Reminder:</b> <br>
This weather dataset is used for technical demonstrations. When thinking about data gathering in your workplace, consider your own sector's data sources—patient records, sales data, service requests, etc.
</div>

### Quantitative Data Gathering Example

**Scenario:** A local council wants to determine optimal timing for outdoor events.

**Quantitative approach:** Analyze historical weather measurements to identify patterns.

````python
# Quantitative analysis: Count rainy days by month
weather_df['date'] = pd.to_datetime(weather_df['date'])
weather_df['month'] = weather_df['date'].dt.month

# Count days with precipitation > 0
rainy_days = weather_df.groupby('month')['precipitation'].apply(
    lambda x: (x > 0).sum()
)

print("Number of rainy days per month:")
print(rainy_days)
````

<div class="alert alert-block alert-info">
<b>What just happened?</b><br>

- We loaded weather data covering multiple years
- We extracted the month from each date
- We counted days with measurable precipitation per month
- The results show objective, numerical evidence of rainfall patterns

**Key lesson:** Quantitative data answers "what" and "how much"—in this case, April shows high rainfall frequency, suggesting the council should avoid scheduling outdoor events then.

</div>

Now let's see what quantitative data **cannot** tell us:

**What the numbers miss:**
- Why do residents prefer certain event types?
- How do cultural factors influence attendance?
- What barriers prevent participation (transport, accessibility, awareness)?

These require **qualitative methods** like resident interviews or focus groups.

## Part 2: Identifying Bias, Noise, and Error

### Visual Overview

```mermaid
flowchart LR
    A[True Reality] --> B[Measurement Process]
    B --> C{Data Collected}
    B -.->|Bias| D[Systematic Distortion]
    B -.->|Noise| E[Random Variation]
    B -.->|Error| F[Inaccurate Values]
    D --> C
    E --> C
    F --> C
    C --> G[Your Analysis]
```

**What this diagram shows:**
Data collection introduces bias (systematic distortion), noise (random variation), and error (inaccuracies). Understanding these helps you evaluate data quality.

**Why this matters in your role:**
In healthcare, a blood pressure cuff consistently reading 5 points high shows bias. Random patient movement adds noise. Recording the wrong patient's reading is error. Each affects data differently.

---

### Worked Example: Detecting Measurement Issues

````python
# Examine temperature sensor data for issues
temps = weather_df['temperature'].dropna()

print(f"Mean temperature: {temps.mean():.1f}°C")
print(f"Standard deviation: {temps.std():.1f}°C")
print(f"Min temperature: {temps.min():.1f}°C")
print(f"Max temperature: {temps.max():.1f}°C")
````

<div class="alert alert-block alert-info">
<b>What just happened?</b><br>

- We calculated summary statistics for temperature readings
- Mean shows the average measurement
- Standard deviation indicates measurement variability (noise)
- Min/max reveal if any readings are implausible (potential errors)

**Key lesson:** If we saw a minimum of -50°C in London, we'd suspect sensor error. If standard deviation was unusually high, we'd investigate noise sources.

</div>

Try modifying the code above to check for potential issues. Examine precipitation or wind speed instead of temperature. Do the ranges seem plausible for London?

### Common Sources of Bias

**Example - Retail:** A supermarket surveys customer satisfaction at 2pm on weekdays. **Bias:** They miss evening and weekend shoppers (working professionals, families), skewing results toward retirees with different needs.

**Example - Healthcare:** A hospital assesses treatment effectiveness by reviewing patient records. **Bias:** Patients who didn't return for follow-up aren't included, potentially hiding poor outcomes.

**Example - Local Government:** A council gauges public opinion via online surveys. **Bias:** Excludes residents without internet access (often elderly or low-income), whose views may differ significantly.

## Part 4: Interactive Practice - Method Selection

Let's practice with a weather-related business scenario.

````python
# Calculate average temperature by season
def get_season(month):
    if month in [12, 1, 2]:
        return 'Winter'
    elif month in [3, 4, 5]:
        return 'Spring'
    elif month in [6, 7, 8]:
        return 'Summer'
    else:
        return 'Autumn'

weather_df['season'] = weather_df['month'].apply(get_season)
seasonal_temps = weather_df.groupby('season')['temperature'].mean()

print("Average temperature by season:")
print(seasonal_temps)
````

<div class="alert alert-block alert-info">
<b>What just happened?</b><br>

- We created a function to categorize months into seasons
- We applied this function to create a new 'season' column
- We calculated mean temperature for each season
- Results show clear seasonal patterns in temperature

**Key lesson:** Grouping and aggregating quantitative data reveals patterns that inform business decisions.

</div>

**Business Context:** A garden center wants to optimize staff scheduling and inventory. This quantitative analysis shows when demand should peak (warmer seasons), but what would qualitative methods add?

**Qualitative additions:**
- **Customer interviews:** Why do some garden in winter (hobby greenhouse users)? What stops summer gardeners (too hot, on holiday)?
- **Staff observations:** Which products generate questions? When do customers seem confused or overwhelmed?
- **Focus groups:** What would improve the shopping experience? What products are customers seeking but not finding?

Numbers show patterns; conversations reveal motivations and opportunities.

**Scenario:** Healthcare - 25% increase in missed GP appointments

**1. Quantitative Method: Appointment System Data Analysis**
- **Data collected:** 6 months of appointment records including booking date, appointment date, patient age group, appointment type, cancellation/no-show status, booking channel (phone/online)
- **Source:** GP practice management system
- **Sample size:** All appointments (~3,000 per month = 18,000 total)
- **Analysis:** Calculate no-show rates by patient demographic, appointment type, time of day, booking lead time
- **Reveals:** *When* and *who* is missing appointments most frequently

**2. Qualitative Method: Patient Interviews**
- **Participants:** 30 patients who missed appointments in the last 3 months
- **Selection:** Mix of demographics, appointment types, booking channels
- **Questions:** 
  - What prevented you from attending or canceling?
  - Are appointment reminders reaching you? Are they helpful?
  - What would make it easier to attend or give notice?
  - Do appointment times fit your work schedule?
- **Reveals:** *Why* patients miss appointments and what barriers exist

**3. Potential Issues:**

*Quantitative (System Data):*
- **Bias:** Doesn't capture why patients didn't cancel (forgot, emergency, transport issues)
- **Error:** Data entry mistakes in demographic fields; system timestamps might be inaccurate
- **Coverage:** Excludes patients who stopped booking entirely due to previous difficulties

*Qualitative (Interviews):*
- **Bias:** Willing participants may not represent all no-shows (hardest-to-reach patients likely underrepresented)
- **Noise:** Individual circumstances vary; difficult to identify common patterns
- **Response bias:** Patients may give socially acceptable answers rather than true reasons

**4. How They Complement:**
- **Quantitative shows patterns:** "No-shows increased 40% for evening appointments" or "Online bookings have 2x higher no-show rate"
- **Qualitative explains causes:** "Evening appointments conflict with school pickup" or "Online booking confirmation emails go to spam"
- **Together inform solutions:** Quantitative identifies which segments to target; qualitative reveals what interventions would work

**5. Expected Business Insights:**
- **Resource allocation:** If analysis shows certain appointment types have higher no-shows, adjust staffing
- **System improvements:** If interviews reveal booking process issues, prioritize UX fixes
