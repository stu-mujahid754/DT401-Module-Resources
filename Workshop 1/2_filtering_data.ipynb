## Part 1: Filtering with WHERE

The `WHERE` keyword is the foundation of all filtering in SQL. It restricts which rows are returned by specifying a condition — only rows where that condition evaluates to `TRUE` are included in the result.

```sql
SELECT column1, column2
FROM table_name
WHERE column_name = value;
```

WHERE always goes after FROM. You will add more keywords after WHERE as queries become more complex, but the position of WHERE stays the same.

### 2.0 Filtering by equality

The simplest filter checks whether a column equals a specific value. For text values, wrap the value in single quotes. For numbers, write the value directly.

## Part 2: Filtering with Comparison Operators

Equality filtering checks for an exact match. Comparison operators let you filter by thresholds — amounts greater than a value, dates before a certain point, or records that do not match a value at all.

### 2.1 Filtering with comparison operators

| Operator | Meaning |
|---|---|
| `>` | Greater than |
| `>=` | Greater than or equal to |
| `<` | Less than |
| `<=` | Less than or equal to |
| `!=` | Not equal to |

```sql
SELECT *
FROM payment
WHERE amount > 5;
```

These operators work on numbers, dates, and text (text is compared alphabetically).

## Part 3: Combining Conditions (AND / OR)

Real queries rarely filter on a single condition. You will often need to combine conditions — for example, customers from a specific store who registered after a certain date, or films that are either short or inexpensive.

### 2.2 Filtering with multiple conditions

Combine conditions with `AND` (both must be true) or `OR` (either can be true):

```sql
SELECT *
FROM customer
WHERE store_id = 1
    AND address_id > 150;
```

Use brackets to control how conditions are grouped — without brackets, AND is evaluated before OR, which can produce unexpected results:

```sql
-- Without brackets: finds store_1 AND MARY, OR any TERRY regardless of store
WHERE store_id = 1 AND first_name = 'MARY' OR first_name = 'TERRY'

-- With brackets: finds MARY or TERRY, but only from store 1
WHERE store_id = 1
    AND (first_name = 'MARY' OR first_name = 'TERRY')
```

When in doubt, add brackets — they make your intent explicit and prevent logic errors.

## Part 4: IN and BETWEEN

AND and OR work well for a small number of conditions, but they become unwieldy when you need to match one of many values or check whether a value falls within a range. IN and BETWEEN handle these cases more concisely.

### 2.3 Filtering by a list (IN)

When you want to match one of several values, `IN` replaces a chain of OR conditions:

```sql
-- These two queries are equivalent:
WHERE first_name = 'MARY' OR first_name = 'TERRY'

WHERE first_name IN ('MARY', 'TERRY')
```

IN becomes particularly valuable when matching longer lists — three OR conditions are manageable, but ten become difficult to read and maintain.

> **Tip:** `IN` can also accept a subquery that returns a single column — this allows you to filter dynamically based on another query's results. You will explore this in [5 — Subqueries and Tables](5_subqueries_and_tables.ipynb).

### 2.4 Filtering by range (BETWEEN)

For ranges of numbers or dates, `BETWEEN ... AND ...` is more concise than writing `>= lower AND <= upper`:

```sql
SELECT *
FROM payment
WHERE amount BETWEEN 5 AND 8.99;
```

BETWEEN is **inclusive on both ends**, so this returns payments of exactly 5.00, exactly 8.99, and everything in between.

## Part 5: Pattern Matching with LIKE

Equality and comparison operators require you to know the exact value you are looking for. LIKE lets you search for patterns — useful when you only know part of a value, or when you want to find all records sharing a common string.

### 2.5 Filtering by pattern (LIKE)

Use `LIKE` with the `%` wildcard to match patterns in text:

| Pattern | Matches |
|---|---|
| `LIKE 'A%'` | Starts with A |
| `LIKE '%a'` | Ends with a |
| `LIKE '%ar%'` | Contains 'ar' anywhere |

```sql
SELECT *
FROM customer
WHERE first_name LIKE 'A%';
```

> **Note:** LIKE is case-sensitive in most databases. Use `UPPER()` to avoid case issues: `WHERE UPPER(first_name) LIKE 'A%'`. You will practise this in Q38 and Q39 below.

Because all names are stored in uppercase in this database, `'%AR%'` will find all matches. In a database with mixed-case names, you would need `UPPER(first_name) LIKE '%AR%'` to be safe.

## Part 6: Handling Missing Values (NULL)

Real datasets are rarely complete. Fields are left blank, records are imported without all their information, or certain data simply was not collected. SQL represents missing values as `NULL` — and handling them requires special syntax.

### 2.6 Filtering by missing values (NULL)

`NULL` means "no value" — it is not the same as zero, an empty string, or the word "null". Because `NULL` represents the absence of a value, standard comparison operators do not work on it:

```sql
-- This does NOT work as expected:
WHERE column_name = NULL

-- Use IS NULL instead:
WHERE column_name IS NULL

-- And IS NOT NULL to find rows with values:
WHERE column_name IS NOT NULL
```

The reason `= NULL` fails is that NULL is not equal to anything — not even to another NULL. `IS NULL` is the correct way to test for absence.
