### The Two Types of Cell

Everything in a Jupyter notebook lives inside a **cell**. There are two types you will use:

**Markdown cells** (like this one)  
Markdown cells contain text, headings, lists, and images. They are used to explain, instruct, and provide context. You are reading a markdown cell right now.

**Code cells**  
Code cells contain Python instructions. When you run a code cell, Python executes the instructions and displays any output directly below the cell.

### Variables

A **variable** is a named container that stores a value. You create a variable by choosing a name, writing `=`, and then providing the value.

```python
city = "London"
temperature = 15.3
is_raining = True
```

Once a variable exists, you can use its name anywhere you would use the value itself. Variables make code easier to read and reuse.

Python has three types of values you will encounter most often:

| Type | Example | Description |
|------|---------|-------------|
| `str` (string) | `"London"` | Text, always in quotes |
| `int` / `float` | `15` / `15.3` | Whole numbers / decimal numbers |
| `bool` | `True` / `False` | Boolean: represents yes/no, on/off |

### Worked Example: Creating Variables

The cell below creates three variables describing a weather observation, then prints them.

````python
# Create some variables
city = "London"
temperature = 15.3       # degrees Celsius
wind_speed = 12          # miles per hour
is_raining = False

# Print each variable
print(city)
print(temperature)
print(wind_speed)
print(is_raining)
````

<div class="alert alert-block alert-info">
<b>What just happened?</b><br>

- We created four variables, each holding a different type of value
- Lines starting with <code>#</code> are <b>comments</b> — Python ignores them, but they explain the code to human readers
- Each <code>print()</code> call displayed the current value stored in that variable
- Notice that text variables display without quotes, while the value <code>False</code> is printed with a capital F (Python is case-sensitive)

<b>Key lesson:</b> Variables let you store values with meaningful names. <code>temperature = 15.3</code> is far more readable than just using <code>15.3</code> repeatedly throughout your code.

### Simple Calculations

Python can perform arithmetic using familiar symbols:

| Operation | Symbol | Example |
|-----------|--------|---------|
| Addition | `+` | `5 + 3` → `8` |
| Subtraction | `-` | `10 - 4` → `6` |
| Multiplication | `*` | `3 * 4` → `12` |
| Division | `/` | `15 / 2` → `7.5` |

You can use variables in calculations, and store the result in a new variable.

### Worked Example: Temperature Conversion

Weather data is often recorded in Celsius. The cell below converts a Celsius temperature to Fahrenheit using the formula: **F = (C × 9/5) + 32**

````python
# Temperature in Celsius
temp_celsius = 18.5

# Convert to Fahrenheit
temp_fahrenheit = (temp_celsius * 9 / 5) + 32

print("Temperature in Celsius:", temp_celsius)
print("Temperature in Fahrenheit:", temp_fahrenheit)
````

<div class="alert alert-block alert-info">
<b>What just happened?</b><br>

- We stored a temperature value in <code>temp_celsius</code>
- We calculated the Fahrenheit equivalent using a formula and stored it in <code>temp_fahrenheit</code>
- We used <code>print()</code> with a comma to display both a label (text) and a value (number) together on one line
- Python follows standard mathematical order of operations: brackets first, then multiplication/division, then addition/subtraction

<b>Key lesson:</b> Storing intermediate results in variables makes calculations easier to follow, check, and reuse.
</div>

### About the Dataset

The file `open-meteo-51.49N0.16W23m.csv` contains hourly weather observations from a location in London (**51.49°N, 0.16°W**, at 23 metres elevation).

The dataset covers **January 2023 to January 2026** — over three years of hourly readings.

Each row in the file represents **one hour** of weather observations. The columns include:

| Column | Description |
|--------|-------------|
| `time` | Date and time of the observation |
| `temperature_2m (°C)` | Air temperature at 2 metres above ground |
| `relative_humidity_2m (%)` | Relative humidity at 2 metres |
| `rain (mm)` | Rainfall in millimetres |
| `wind_speed_10m (mp/h)` | Wind speed at 10 metres above ground |
| `wind_speed_100m (mp/h)` | Wind speed at 100 metres (turbine height) |
| `cloud_cover (%)` | Total cloud cover percentage |
| `sunshine_duration (s)` | Seconds of sunshine in that hour |
| `precipitation (mm)` | Total precipitation (rain + snow) |

...and many more columns covering soil moisture, pressure, dew point, and more.

<div class="alert alert-block alert-warning">
<b>Reminder — Data, Dataset, and Database:</b> <br>
In DT401.1, you explored the distinction between <b>data</b>, <b>datasets</b>, and <b>databases</b>:

<ul>
<li>A single temperature reading — e.g. *11.2°C at midnight on 1 January 2023* — is a piece of <b>data</b></li>
<li>All 27,000+ hourly readings together form a <b>dataset</b></li>
<li>If this data were stored in an organised system alongside other related data (rainfall history, forecasts, sensor metadata), it would form part of a <b>database</b></li>
</ul>

The CSV file we are about to load is a dataset — structured, organised, and ready for analysis.
</div>

### Worked Example: Loading the Dataset

We use `pd.read_csv()` to load the file. This creates a **DataFrame** — pandas' name for a data table with rows and columns.

One thing to know about this particular file: the first three rows contain location metadata (latitude, longitude, elevation) rather than data. We use `skiprows=3` to skip those and start reading from the actual column headers.

````python
import pandas as pd

# Load the METEO weather dataset
# skiprows=3 skips the location metadata at the top of the file
df = pd.read_csv('../open-meteo-51.49N0.16W23m.csv', skiprows=3)

# Keep only rows with valid timestamps (removes any stray blank rows)
df = df[df['time'].str.match(r'^\d{4}-\d{2}-\d{2}T', na=False)].reset_index(drop=True)

print(f"Dataset loaded successfully.")
print(f"Total rows: {len(df):,}")
print(f"Time range: {df['time'].iloc[0]} to {df['time'].iloc[-1]}")
````

<div class="alert alert-block alert-info">
<b>What just happened?</b><br>

- <code>pd.read_csv()</code> opened the file and loaded it into a pandas DataFrame called <code>df</code>
- <code>skiprows=3</code> told pandas to ignore the first 3 rows (location information, not measurements)
- The second line filtered out any blank rows that might appear at the end of the file
- We used <code>len(df)</code> to count the number of rows and <code>df['time'].iloc[0]</code> to read the first timestamp
- The format <code>f"...{variable}..."</code> is called an <b>f-string</b> — it lets you embed variable values directly inside printed text

<b>Key lesson:</b> Real-world data files often contain metadata or formatting that requires a small amount of preparation before analysis. <code>skiprows</code> is a common solution for CSV files with header information.
</div>

### Accessing a Single Column

To work with one specific column, use the column name in square brackets:

```python
df['column_name']
```

This returns a **Series** — a single column of data. You can then apply methods to it, such as `.mean()`, `.max()`, `.min()`, and `.sum()`.

### Worked Example: Temperature Column

````python
# Access the temperature column
temperature_column = df['temperature_2m (°C)']

# Calculate basic statistics
print("Temperature column — summary statistics")
print(f"  Average : {pd.to_numeric(temperature_column, errors='coerce').mean():.1f} °C")
print(f"  Highest : {pd.to_numeric(temperature_column, errors='coerce').max():.1f} °C")
print(f"  Lowest  : {pd.to_numeric(temperature_column, errors='coerce').min():.1f} °C")
````

<div class="alert alert-block alert-info">
<b>What just happened?</b><br>

- We extracted the <code>temperature_2m (°C)</code> column into a new variable called <code>temperature_column</code>
- We applied three methods to it: <code>.mean()</code>, <code>.max()</code>, and <code>.min()</code>
- The <code>:.1f</code> inside the f-string formats the output to 1 decimal place
- Each method returns a single number — a summary of all 27,000+ values in that column

<b>Key lesson:</b> Isolating a column lets you focus your analysis. You do not have to extract to a variable — you can also write <code>df['temperature_2m (°C)'].mean()</code> directly.
</div>

### Worked Example: A Simple Question

Let us ask a simple question of our data: **How many hours in this dataset had some form of precipitation?**

This is a basic form of business intelligence — turning raw data into an answer.

````python
# Count hours with precipitation
hours_with_rain = (df['precipitation (mm)'] > 0).sum()
total_hours = len(df)
rain_percent = (hours_with_rain / total_hours) * 100

print(f"Total hours in dataset : {total_hours:,}")
print(f"Hours with precipitation: {hours_with_rain:,}")
print(f"Percentage of wet hours : {rain_percent:.1f}%")
````

<div class="alert alert-block alert-info">
<b>What just happened?</b><br>

- <code>df['precipitation (mm)'] > 0</code> tests each value: is it greater than zero? This produces a column of <code>True</code> / <code>False</code> values
- <code>.sum()</code> counts all the <code>True</code> values — because Python treats <code>True</code> as <code>1</code> and <code>False</code> as <code>0</code>
- We divided the count by the total rows to get a percentage

This one-line pattern — <code>(condition).sum()</code> — is extremely useful in data analysis. You will use it frequently.

<b>Think about it:</b> This single number, derived from 27,000+ readings, could inform a real business decision. A facility manager planning outdoor events, a logistics company scheduling deliveries, or an insurer modelling weather risk — all might ask exactly this question.
